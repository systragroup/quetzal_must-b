# Distribution factors

```python
zones = gpd.read_file('../../inputs/zones/raw_fine/fr_adm_commune_local.gpkg').to_crs(epsg=4326)
```

```python
zones['zone'] = zones['zone'].fillna('HBM')
zones.groupby('zone')['population'].sum()

coeff = 1 - zones.loc[zones['zone'] == 'BIS', 'population'].sum()/zones.loc[zones['zone'] == 'HBM', 'population'].sum()
print(coeff)
```

```python
### Totaux de déplacements (EMC2 2021)
int_BM = 2525000
int_HBM = 2148000
exch_BM_HBM = 483000

nb_deps = int_BM + coeff*(int_HBM + exch_BM_HBM)
print(nb_deps)
```

```python
### Ramener à une heure à la pointe:
### - 50% du traffic sur une préiode de pointe de 5 heures (source EMC2)
### ---> donc 10% du traffic sur une heure de pointe

nb_deps_HP = nb_deps/10

print(nb_deps_HP)
```
